# Relation Cardinality Analysis

This notebook is a lightweight viewer for relation-cardinality artifacts generated by scripts.
It should not contain hidden experiment logic.

Required command:

```bash
PYTHONPATH=src .venv/bin/python scripts/structural_analysis.py
PYTHONPATH=src .venv/bin/python scripts/router_analysis.py --skip-router
```

The final full comparison figure includes `AnyBURL`, `ConvE`, `HoGRN`, `PathBSR`, and `TransE` on all six datasets.

````python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display, Markdown

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "pyproject.toml").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
RESULTS = PROJECT_ROOT / "results"

MODEL_ORDER = ["AnyBURL", "ConvE", "HoGRN", "PathBSR", "TransE"]
DATASET_ORDER = ["FB15K-237-10", "FB15K-237-20", "FB15K-237-50", "NELL23K", "WD-singer", "WN18RR"]
````

````python
summary_path = RESULTS / "cardinality/all_datasets_cardinality_mrr_by_model.csv"
figure_path = RESULTS / "cardinality/all_datasets_cardinality_mrr_by_model.png"
if not summary_path.exists() or not figure_path.exists():
    display(Markdown("""
**Missing generated artifact.** Run:

```bash
PYTHONPATH=src .venv/bin/python scripts/structural_analysis.py
PYTHONPATH=src .venv/bin/python scripts/router_analysis.py --skip-router
```
"""))
else:
    df = pd.read_csv(summary_path)
    df["dataset"] = pd.Categorical(df["dataset"], DATASET_ORDER, ordered=True)
    df["model"] = pd.Categorical(df["model"], MODEL_ORDER, ordered=True)
    display(df.sort_values(["dataset", "rel_type", "model"]).head())
    display(Image(filename=str(figure_path)))
````

````python
# Full cardinality MRR table: dataset x relation-cardinality bucket x model.
if summary_path.exists():
    pivot = (
        pd.read_csv(summary_path)
        .pivot_table(index=["dataset", "rel_type"], columns="model", values="mrr", aggfunc="first")
        .reindex(columns=MODEL_ORDER)
    )
    display(pivot.round(4))
````

````python
# Query-share table for the FB sparsity splits.
share_path = RESULTS / "cardinality/fb_cardinality_query_share.csv"
if share_path.exists():
    display(pd.read_csv(share_path).round(3))
````

````python
# Existing paper-focused cardinality figures.
for name in [
    "fb_cardinality_mrr_grouped_bar.png",
    "nell_wd_cardinality_mrr.png",
]:
    path = RESULTS / "cardinality" / name
    if path.exists():
        display(Markdown(f"### `{name}`"))
        display(Image(filename=str(path)))
````
